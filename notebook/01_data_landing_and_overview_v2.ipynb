```python
fields = ['posisi_pekerjaan_clean']

df = pd.read_csv('../datasets/dataset_clean.csv', skipinitialspace=True, usecols=fields)

job_counts = df['posisi_pekerjaan_clean'].value_counts()
print(job_counts)

total_unique = df['posisi_pekerjaan_clean'].nunique()
print(f"Total unique job positions: {total_unique}")
```

```python
job_table = df['posisi_pekerjaan_clean'].value_counts().head(12).reset_index()
job_table.columns = ['Nama Role', 'Total Sample']

job_table['Persentase'] = (job_table['Total Sample'] / len(df) * 100).round(2).astype(str) + '%'

job_table
```
