```python
import pandas as pd
import re

# Load dataset
df = pd.read_csv('../datasets/dataset.csv')
print(f"Total baris data: {len(df)}")
df.head()
```

```python
def normalize_job_title(title):
    if pd.isna(title):
        return "Unknown"

    t = str(title).strip().lower()

    if 'fullstack' in t or 'full stack' in t or 'full-stack' in t:
        return 'Fullstack Engineer'
    elif 'frontend' in t or 'front end' in t or 'front-end' in t:
        return 'Frontend Engineer'
    elif 'backend' in t or 'back end' in t or 'back-end' in t:
        return 'Backend Engineer'
    elif 'qa' in t or 'quality assurance' in t or 'tester' in t:
        return 'QA / Software Tester'
    elif 'devops' in t or 'sre' in t or 'site reliability' in t:
        return 'DevOps / SRE'
    elif 'data scientist' in t:
        return 'Data Scientist'
    elif 'data engineer' in t:
        return 'Data Engineer'
    elif 'machine learning' in t or 'ml' in t or 'ai engineer' in t:
        return 'Machine Learning / AI Engineer'
    elif 'cyber' in t or 'security' in t:
        return 'Cybersecurity Specialist'
    elif 'ui' in t or 'ux' in t or 'product design' in t:
        return 'UI/UX Designer'
    elif 'product manager' in t or 'pm' in t:
        return 'Product Manager'
    elif 'technical writer' in t or 'writer' in t:
        return 'Technical Writer'
    else:
        return title.title()

df['posisi_pekerjaan_clean'] = df['posisi_pekerjaan'].apply(normalize_job_title)

print("=== Distribusi Posisi Pekerjaan Setelah Normalisasi ===")
print(df['posisi_pekerjaan_clean'].value_counts())

output_filename = '../datasets/dataset_clean.csv'
df.to_csv(output_filename, index=False)
print(f"\n[SUKSES] Dataset baru telah disimpan sebagai '{output_filename}'")
```
